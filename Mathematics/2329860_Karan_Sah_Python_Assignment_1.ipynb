```python
#A function written to calculate BMR for male and female using ‘Harris-Benedict’ method
def calculate_bmr(gender, weight_kg, height_cm, age_years):
    if gender == "men":
        bmr = 88.362 + (13.397 * weight_kg) + (4.799 * height_cm) - (5.677 * age_years)
    elif gender == "women":
        bmr = 447.593 + (9.247 * weight_kg) + (3.098 * height_cm) - (4.330 * age_years)
    else:
        raise ValueError("Invalid gender specified. Please specify 'men' or 'women'.")

    return bmr


# A function written to calcualte recommended daily kilocalorie intake on the basis of exercises performed.
def calculate_daily_intake(bmr, exercise_level):
    if exercise_level == 0:
        daily_intake = bmr * 1.2
    elif exercise_level == 1:
        daily_intake = bmr * 1.375
    elif exercise_level == 2:
        daily_intake = bmr * 1.55
    elif exercise_level == 3:
        daily_intake = bmr * 1.725
    elif exercise_level == 4:
        daily_intake = bmr * 1.9
    else:
        raise ValueError("Invalid exercise level specified. Please specify a number from 0 to 4.")

    return daily_intake


# Welcome messages and short description on exercises level
print("Welcome to Basal Metabolic Rate and Daily Required KC Intake Calculator using ‘Harris-Benedict’ Method")
print("Exercises Levels")
print("0	 Little to no exercise ")
print("1	 Light exercise (1–3 days per week)")
print("2	Moderate exercise (3–5 days per week)")
print("3	 Heavy exercise (6–7 days per week)	")
print("4	 Very heavy exercise (twice per day, extra heavy workouts)")
# Prompts the user for their input 
gender = input("Enter your gender (men/women): ")
weight_kg = float(input("Enter your weight in kilograms: "))
height_cm = float(input("Enter your height in centimeters: "))
age_years = int(input("Enter your age in years: "))
exercise_level = int(input("Enter your exercise level (0-4): "))

# BMR and daily recommend calorie intake calculation is done
bmr = calculate_bmr(gender, weight_kg, height_cm, age_years)
daily_intake = calculate_daily_intake(bmr, exercise_level)

# The results are now displayed to the user.
print("Your Basal Metabolic Rate (BMR) is:", bmr, "kilocalories per day.")
print("Your recommended daily kilocalorie intake is:", daily_intake, "kilocalories per day.")
```

Demonstrate this program works with:

```python
#to calulate the sum of values between the first and last student number
def sum_between(low, high):
    # to check whether the inputs are integers and if they are valid
    if not isinstance(low, int) or not isinstance(high, int):
        print("Invalid input: low and high must be integers.")
        return None
    #  to check whether low value is equal to or less than the high value
    if low > high:
        print("Invalid input: low value cannot be greater than high value.")
        return None
    # base case: if low and high are equal, return their value
    if low == high:
        return low
    # recursive case: sum the current low value and recursively sum the rest
    return low + sum_between(low+1, high)

# prompt and ask the users for their input
low = input("Enter the low number: ")
high = input("Enter the high number: ")

# converting the inputs to integers
try:
    low = int(low)
    high = int(high)
except ValueError:
    print("Invalid input: low and high must be integers.")
else:
    # call sum_between with the user inputs
    result = sum_between(low, high)
    if result is not None:
        print(f"The sum between {low} and {high} is {result}.")
```

Create a Py Program that will create a graph of y = f(x) where; 

f(x) = 3x2  

And where

f(x) = 4x - 3

Your graph will need to plot the values of y for the values of x between -5 and +15 for both lines.

To completely address this question you will need to; 
●	Correctly label the created graph with a title and labelled axes.
●	Use a function to calculate the values of y for a given x value for both lines.
●	Use plot markers for the graphs plotted points for both lines. 
●	Include a ‘grid’ on the graph.
●	Present as evidence your code and the created graph with both lines visible.


```python
#importing libraries
import matplotlib.pyplot as plt
import numpy as np
#creating function for f(x) = 3 * x**2
def function1(x):
    return 3 * x**2
#creating function for f(x) = 4 * x - 3
def function2(x):
    return 4 * x - 3

x = np.linspace(-5, 15, 100) # 100 values between -5 and 15
y1 = function1(x)
y2 = function2(x)
#creating plot markers 
plt.plot(x, y1, 'o-', label='f(x) = 3x^2')
plt.plot(x, y2, 'o-', label='f(x) = 4x - 3')
#labeling the graph and labeling the axes
plt.title('Graph of y=f(x)')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
#including a gird on the graph 
plt.grid(True)
plt.legend()
plt.show()
```
