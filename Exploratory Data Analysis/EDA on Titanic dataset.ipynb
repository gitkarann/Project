Data Cleaning

```python
# Impute missing values for 'Age' and 'Fare' with median
df['Age'].fillna(df['Age'].median(), inplace=True)
df['Fare'].fillna(df['Fare'].median(), inplace=True)
```

Median method is used for data imputation in this case as it is quic way to fill missing values as the values in age and fare are missing at random (MAR).

```python
# Impute missing values for 'Embarked' with mode
mode_embarked = df['Embarked'].mode()[0]
df['Embarked'].fillna(mode_embarked, inplace=True)
```

Mode method is used for data imputation in this case as this is categorical variable(take on one of a limited and usually fixed number of values).

```python
# Drop the 'Cabin' variable
df.drop(columns=['Cabin'], inplace=True)
```

As the missing values of cabin column is as high as 77.46 % and it has signigicant part of data missing, filling up these values may cause uncertatinty and bais in the analysis.

Vizualize the Data

Univariate Analysis:

```python

# Set the style of seaborn
sns.set(style="whitegrid")

# Plotting the distributions and outliers
fig, axes = plt.subplots(nrows=3, ncols=2, figsize=(14, 12))

# Line chart for Age distribution
sns.histplot(df['Age'], kde=True, color='skyblue', ax=axes[0, 0])
axes[0, 0].set_title('Distribution of Age')
axes[0, 0].set_xlabel('Age')

# Boxplot for Age to analyze outliers
sns.boxplot(x=df['Age'], ax=axes[0, 1], color='lightcoral')
axes[0, 1].set_title('Boxplot of Age')
axes[0, 1].set_xlabel('Age')

# Bar chart for Fare distribution
sns.histplot(df['Fare'], kde=True, color='salmon', ax=axes[1, 0])
axes[1, 0].set_title('Distribution of Fare')
axes[1, 0].set_xlabel('Fare')

# Boxplot for Fare to analyze outliers
sns.boxplot(x=df['Fare'], ax=axes[1, 1], color='lightgreen')
axes[1, 1].set_title('Boxplot of Fare')
axes[1, 1].set_xlabel('Fare')

# Pie chart for Sex distribution
sex_counts = df['Sex'].value_counts()
axes[2, 0].pie(sex_counts, labels=sex_counts.index, autopct='%1.1f%%', colors=['lightblue', 'lightcoral'])
axes[2, 0].set_title('Distribution of Sex')

# Pie chart for Embarked distribution
embarked_counts = df['Embarked'].value_counts()
axes[2, 1].pie(embarked_counts, labels=embarked_counts.index, autopct='%1.1f%%', colors=['lightgreen', 'lightcoral', 'skyblue'])
axes[2, 1].set_title('Distribution of Embarked')

plt.tight_layout()
plt.show()
```

Interpretation:

*   Distribution of Age: The line chart depicts the distribution of ages of passengers ages which shows the desnity of value of age. Boxplot summarizes central tendecy spread an dpresence of outliers of the age distribution. Age distribution is slightly right-skewed and has peak in young adult age.

*   Distribution of Fare: The bar chart shows frequency of fare values. The boxplot summarizes central tendency and potential outliers of fare distribution. The majority of passengers paid lower fares thus right-skewed. Boxplot show the presence of less high outliers which means few paid relatively larger fares.


*   Distribution of Sex: Pie chart shows there were more males compared to females on the Titanic which is shown by larger blue cut.

*   Distribution of Embarked: The pie chart depicts embarkation ports, largest segment being from Southampton(S) then Cherbourg(C) comes second. Nevertheless, there were fewest passengers from Queenstwon(Q).





Bi-variate Analysis

```python

# Separate numeric and categorical columns
numeric_columns = df.select_dtypes(include='number')
categorical_columns = df.select_dtypes(exclude='number')

# 1. Explore correlation visualization with scatter plot between the variables (Pairplot)
sns.pairplot(numeric_columns)
plt.suptitle('Pairplot of Numeric Variables', y=1.02)
plt.show()

# 2. Explore the correlation visualization with the heatmap
correlation_matrix = numeric_columns.corr()
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Heatmap of Numeric Variables')
plt.show()
```

Intrepretation:

*  Pairplot of Numeric Variables:
The pairplot shows scatter plots for every pair of variable of numericals in the dataset. Diagonal elements shows the distribution of single variables. It allows to inspect correlations and relations between numeric variables.
It shows how the variables are related. In the plot age vs. fare there is not any linear relation between age and fare. Individual characters is shown by diagonal histogram of each variable.
*   Correlation Heatmap of Numeric Variables:
Represents the correlation matirix of numeric variables. Colors depicts the strength and direction of correlation between variables. Darker colors show stronger positive correlations and lighter color mean negative correlation. Pclass and Fare have strong correlation. This provide a quantitaive view of correlation.

The reader can distinguish the variables strongly correlated and find patterns and dependencies. Correlation can guide feature selection.
